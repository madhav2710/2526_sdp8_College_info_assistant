```python
import os
from google.colab import userdata
from supabase.client import create_client, Client
from langchain_google_genai import GoogleGenerativeAIEmbeddings, ChatGoogleGenerativeAI
from langchain_community.vectorstores import SupabaseVectorStore
from langchain_core.documents import Document
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# --- 0. SETUP ---
os.environ["GOOGLE_API_KEY"] = userdata.get('GOOGLE_API_KEY')
SUPABASE_URL = userdata.get('SUPABASE_URL')
SUPABASE_KEY = userdata.get('SUPABASE_KEY')

supabase: Client = create_client(SUPABASE_URL, SUPABASE_KEY)

# Initialize Models
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/text-embedding-004",
    task_type="retrieval_document"
)

llm = ChatGoogleGenerativeAI(
    model="gemini-flash-latest",
    temperature=0
)

vector_store = SupabaseVectorStore(
    client=supabase,
    embedding=embeddings,
    table_name="documents",
    query_name="match_documents"
)





```

```python
# --- 1. UPLOAD EMBEDDINGS (If needed) ---
# We only add data if the table is empty to avoid duplicates for this demo
print("--- Step 1: Checking/Uploading Data ---")
docs = [
    Document(page_content="Gemini 1.5 Flash is designed for high-frequency tasks where speed is critical.", metadata={"source": "tech_spec"}),
    Document(page_content="The context window of Gemini 1.5 Pro is 1 million tokens, allowing for massive data analysis.", metadata={"source": "tech_spec"}),
    Document(page_content="Supabase provides a Postgres database with the pgvector extension pre-installed.", metadata={"source": "db_spec"}),
]
# Uncomment the line below to upload fresh data
vector_store.add_documents(docs)
print("Data check complete.\n")
```

```python

# --- 2. EXPLICIT RETRIEVAL (The "Search" Part) ---
# Here we don't use the LLM yet. We just ask the database:
# "Give me the text mathematically closest to this question."
query = "How big is the context window?"
print(f"--- Step 2: Retrieving context for: '{query}' ---")

# k=2 means "Get me the top 2 most similar chunks"
similar_docs = vector_store.similarity_search_with_relevance_scores(query, k=2)

retrieved_context = []

print("Found the following relevant chunks:")
for doc, score in similar_docs:
    print(f"\n[Similarity Score: {score:.4f}]")
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}")

    # Collect the content to pass to the LLM later
    retrieved_context.append(doc.page_content)

# Join the list of strings into one big block of text
context_block = "\n\n".join(retrieved_context)
```

```python
# --- 3. GENERATION (The "Answer" Part) ---
# Now we manually construct the prompt with the text we found above.
print(f"\n--- Step 3: Prompting Gemini with retrieved context ---")

template = """You are a helpful assistant. Use only the following context to answer the question.

Context from database:
{context}

Question:
{question}
"""

prompt = ChatPromptTemplate.from_template(template)
chain = prompt | llm | StrOutputParser()

# Invoke the chain with the explicitly retrieved context
final_answer = chain.invoke({
    "context": context_block,
    "question": query
})

print(f"Gemini's Answer:\n{final_answer}")
```
